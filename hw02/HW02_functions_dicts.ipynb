## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
def to_num(s):
    return float(s.replace(" ", "").replace(",", "."))
def clean_city(s):
    cities = {
        "Almaty": "Алматы",
        "Astana": "Астана",
        "Shymkent": "Шымкент",
        "Aktobe": "Актобе",
        "Atyrau": "Атырау",
        "Karaganda": "Караганда"
    }
    s = s.strip().title()
    return cities.get(s, s)
````

````python
counts = {}
for order in rows:
        category = order["category"]
        counts[category] = counts.get(category, 0) + 1
print(sorted(counts.items(), key=lambda x: x[1], reverse=True))
````

````python
sum_category = {}
avg_category = {}
counts = {}
for order in rows:
        category = order["category"]
        amount = to_num(order["amount"])
        sum_category[category] = sum_category.get(category, 0) + amount
        counts[category] = counts.get(category, 0) + 1
        avg_category[category] = sum_category[category] / (counts[category])
for category in sum_category:
    print(
        f"{category}: "
        f"сумма - {sum_category[category]:.2f}, "
        f"средний чек - {avg_category[category]:.2f}"
    )
````

````python
counts_month = {}
for order in rows:
    month = order["order_date"][:7]
    counts_month[month] = counts_month.get(month, 0) + 1
print(sorted(counts_month.items()))
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
revenue_month = {}
for order in rows:
    if order["status"] == "оплачен":
        month = order["order_date"][:7]
        amount = to_num(order["amount"])
        revenue_month[month] = revenue_month.get(month, 0) + amount
for month, amount in sorted(revenue_month.items()):
    print(f"{month}: {amount:.2f}")
september = revenue_month.get("2026-09", 0)
october = revenue_month.get("2026-10", 0)
if september != 0:
    change = (october - september) / september * 100
    print(f"Изменение: {change:.1f}%")
````

````python
total_city = {}
cancelled_city = {}
for order in rows:
    city = clean_city(order["city"])
    total_city[city] = total_city.get(city, 0) + 1
    if order["status"] == "отменён":
        cancelled_city[city] = cancelled_city.get(city, 0) + 1
for city in sorted(total_city):
    percent = cancelled_city.get(city, 0) / total_city[city] * 100
    print(f"{city}: {percent:.1f}%")
````

````python
revenue_customer = {}
for order in rows:
    if order["status"] == "оплачен":
        amount = to_num(order["amount"])
        customer = order["customer_id"]
        revenue_customer[customer] = revenue_customer.get(customer, 0) + amount
top_5 = sorted(revenue_customer.items(),key=lambda x: x[1],reverse=True)[:5]
for customer, amount in top_5:
    print(f"{customer!r}: {amount:.2f}")
````

Здесь есть тестовый айди пользователь с суммой заказа на 29 999 970

````python
customers_city = {}
for order in rows:
    city = clean_city(order["city"])
    customer = order["customer_id"]
    if city not in customers_city:
        customers_city[city] = set()
    customers_city[city].add(customer)
for city in sorted(customers_city):
    print(f"{city}: {len(customers_city[city])}")
````

## [Челлендж]
9. Найдите полные дубли строк: превратите каждую строку в `tuple(r.values())` и проверяйте через множество. Сколько их и в каком месяце?
10. Пересчитайте выручку по месяцам без дублей и без клиента из задачи 7. Как изменился процент из задачи 5?
11. У какого клиента больше всего заказов (после удаления дублей)? Сколько?
12. Напишите менеджеру 3–4 предложения: насколько на самом деле упала выручка и почему первый расчёт был неверным.

````python
# Челлендж
seen = set()
clean_rows = []
duplicates = {}
for r in rows:
    t = tuple(r.values())
    if t in seen:
        month = r["order_date"][:7]
        duplicates[month] = duplicates.get(month, 0) + 1
    else:
        seen.add(t)
        clean_rows.append(r)
print(duplicates)
````

````python
revenue = {}
for r in clean_rows:
    if r["status"] == "оплачен" and r["customer_id"] != "TEST":
        month = r["order_date"][:7]
        revenue[month] = revenue.get(month, 0) + to_num(r["amount"])
print(revenue)
sep = revenue["2026-09"]
oct = revenue["2026-10"]
change = (oct - sep) / sep * 100
print(f"Изменение: {change:.2f}%")
````

````python
counts = {}
for r in clean_rows:
    customer = r["customer_id"]
    counts[customer] = counts.get(customer, 0) + 1
print(sorted(counts.items(), key=lambda x: x[1], reverse=True)[:1])
````
